```python
path_schema = '/content/drive/MyDrive/schema.csv'
df_schema = pd.read_csv(path_schema, encoding='utf-8')

# Завантажуємо результати
path_results = '/content/drive/MyDrive/results.csv'
df_results = pd.read_csv(path_results, encoding='utf-8')

print("Обидва файли успішно завантажено з Google Диску!")
print(f"Формат схеми (df_schema): {df_schema.shape}")
print(f"Формат відповідей (df_results): {df_results.shape}")
# Витягуємо унікальні назви питань з поля qname, ігноруючи пусті значення
schema_questions = set(df_schema['qname'].dropna().unique())

print(f"Отримано питань із поля 'qname': {len(schema_questions)}")
# Отримуємо множину всіх колонок, які є у файлі відповідей
results_columns = set(df_results.columns)

# Використовуємо інтерсекцію (перетин) двох множин для фільтрації
target_columns = list(schema_questions.intersection(results_columns))

print(f"Кількість колонок після фільтрації (інтерсекції): {len(target_columns)}")
# Залишаємо в таблиці результатів лише відфільтровані колонки
df_filtered_questions = df_results[target_columns]

# Видаляємо респондентів (рядки), у яких є бодай один пропуск (NaN) у цих колонках
df_no_missing = df_filtered_questions.dropna()

# Отримуємо фінальну кількість
final_respondents_count = len(df_no_missing)

print(f"Фінальний результат: {final_respondents_count}")
#У результаті суворої фільтрації за допомогою .dropna() ми отримали 0 респондентів. Отже всі респонденти пропустили хоча б одне питання. Проте аналіз даних показує, що значення NaN
#у деяких колонках (наприклад, про утриманців) логічно означають відсутність ознаки, а не ігнорування питання респондентом.
#Це специфіка проектування анкети опитування
```

```python
#Завдання 3. Статистичний аналіз досвіду респондентів
# Перевіряємо, чи є колонка 'WorkExp' у нашому датасеті
column_name = 'WorkExp'

if column_name in df_results.columns:
    #Обчислюємо середнє значення (mean)
    mean_val = df_results[column_name].mean()

    #Обчислюємо медіану (median)
    median_val = df_results[column_name].median()

    #Обчислюємо моду (mode)
    # Метод .mode() повертає Series (бо мод може бути кілька), беремо перше значення [0]
    mode_val = df_results[column_name].mode()[0]

    #Рахуємо кількість пропущених значень
    missing_count = df_results[column_name].isna().sum()

    # Формуємо красиву підсумкову таблицю (Series)
    central_tendency = pd.Series({
        "Середнє значення (Mean)": round(mean_val, 2),
        "Медіана (Median)": median_val,
        "Мода (Mode)": mode_val
    }, name="Міри центральної тенденції для WorkExp")

    # Виведення результатів
    print(f"Аналіз колонки '{column_name}':")
    print(f"Кількість пропущених значень (NaN), які були проігноровані: {missing_count}\n")
    print(central_tendency)

else:
    print(f"Помилка! Колонку '{column_name}' не знайдено в df_results. Доступні колонки: {list(df_results.columns)[:10]}")
#В середньому досвід роботи - 13 років.
#Більшість респондентів мають досвід роботи - 10 років.
#Медіана - 10 років, отже 50% респондентів мають досвід 10 років і менше, інші 50% — 10 років і більше.
#Середнє значення (13.37) є більшим за медіану. Це вказує на те, що розподіл має правосторонню асиметрію.
#У вибірці є невелика кількість респондентів із дуже великим досвідом (наприклад, 30–40 років), які «тягнуть» середнє значення вгору, тоді як більшість зосереджена навколо 10 років.
```

```python
#Завдання 4. Аналіз віддаленої роботи
# Перевіряємо, чи є колонка 'RemoteWork' у нашому датасеті
column_name = 'RemoteWork'

if column_name in df_results.columns:
    print("Всі доступні варіанти форматів роботи в датасеті:")
    print(df_results[column_name].value_counts(dropna=False))
    print("-" * 50)

    # ВИПРАВЛЕНО: Використовуємо точне порівняння '==', щоб виключити гібридні формати
    df_remote = df_results[df_results[column_name] == 'Remote']

    # Підраховуємо фінальну кількість респондентів
    remote_count = len(df_remote)

    # Фінальний результат
    print(f"Фінальний результат: Кількість респондентів, які працюють віддалено — {remote_count}")

else:
    print(f"Помилка! Колонку '{column_name}' не знайдено. Перевірте список колонок: {list(df_results.columns)[:15]}")
#Якщо фільтрувати суворо лише повністю віддалених працівників (Remote), фінальний результат становить: 10 931 респондент.
```

```python
#Завдання 5. Визначення популярності Python
#Визначаємо правильну колонку з мовами програмування
column_name = 'LanguageHaveWorkedWith'

if column_name in df_results.columns:
    #Рахуємо від УСІХ респондентів у базі
    total_responded = len(df_results)

    # Очищуємо для пошуку тексту, але знаменник (total_responded) не змінюємо
    df_langs = df_results[column_name].dropna()

    #Точний пошук слова Python
    python_users = df_langs.str.contains(r'\bPython\b', na=False, regex=True)
    python_count = python_users.sum()

    #Розрахунок у відсотках
    python_percentage = (python_count / total_responded) * 100

    print(f"Фінальний результат: {python_percentage:.1f}%")
```

```python
#Завдання 6. Аналіз шляхів навчання програмуванню
#Визначаємо правильну колонку зі способами навчання
column_name = 'LearnCode'

if column_name in df_results.columns:
    # Дивимося на унікальні комбінації відповідей (для самоперевірки)
    df_learn = df_results[column_name].dropna()

    #Фільтруємо респондентів, які обрали онлайн-курси
    #В опитуванні Stack Overflow цей варіант зазвичай містить фразу "Online Courses"
    online_courses_users = df_learn.str.contains('Online Courses', na=False, case=False)

    #Підраховуємо результат
    online_courses_count = online_courses_users.sum()

    # Фінальний результат
    print(f"Аналіз способів навчання:")
    print(f"Всього респондентів відповіли на питання 'LearnCode': {len(df_learn)}")
    print(f"Фінальний результат: Кількість респондентів, які навчалися через онлайн-курси — {online_courses_count}")

else:
    print(f"Помилка! Колонку '{column_name}' не знайдено. Перевірте список колонок: {list(df_results.columns)[:20]}")
```

```python
#Завдання 7. Географічний аналіз компенсації Python-розробників

#Визначаємо назви необхідних колонок
lang_col = 'LanguageHaveWorkedWith'
country_col = 'Country'
comp_col = 'ConvertedCompYearly'

# Перевіряємо наявність усіх трьох колонок у датасеті
required_cols = [lang_col, country_col, comp_col]
if all(col in df_results.columns for col in required_cols):

    #Фільтруємо лише Python-розробників
    #Очищуємо від NaN у колонці мов та компенсації для точного аналізу
    df_python = df_results[
        df_results[lang_col].str.contains('Python', na=False) &
        df_results[comp_col].notna()
    ]

    #Групуємо по країнах та обчислюємо середнє (mean) і медіану (median)
    #Функція .agg() дозволяє порахувати кілька показників одночасно
    df_compensation = df_python.groupby(country_col)[comp_col].agg(['mean', 'median'])

    #Перейменовуємо колонки для красивого та зрозумілого вигляду
    df_compensation = df_compensation.rename(columns={
        'mean': 'Середня компенсація',
        'median': 'Медіанна компенсація'
    })

    # Сортуємо за медіанною компенсацією від більшої до меншої (опціонально, для наочності)
    df_compensation = df_compensation.sort_values(by='Медіанна компенсація', ascending=False)

    # Скидаємо індекс, щоб країна стала звичайною колонкою, як вимагає ТЗ
    df_final_table = df_compensation.reset_index()

    # Округляємо значення до 2 знаків після коми
    df_final_table['Середня компенсація'] = df_final_table['Середня компенсація'].round(2)
    df_final_table['Медіанна компенсація'] = df_final_table['Медіанна компенсація'].round(2)

    # 5. Виведення результату (перші 15 країн)
    print("Географічний аналіз компенсації Python-розробників (Top-15 країн за медіаною):")
    display(df_final_table.head(15))

else:
    print("Помилка! Перевірте наявність усіх колонок:", required_cols)
```

```python
#Завдання 8. Аналіз освіти найбільш оплачуваних спеціалістів

#Визначаємо назви колонок для компенсації та освіти
comp_col = 'ConvertedCompYearly'
edu_col = 'EdLevel'

# Перевіряємо наявність обох колонок у датасеті
if comp_col in df_results.columns and edu_col in df_results.columns:

    #Очищуємо дані від пропущених значень у колонці компенсації
    df_clean_comp = df_results[df_results[comp_col].notna()]

    #Сортуємо дані за компенсацією у спадному порядку (ascending=False)
    # та вибираємо топ-5 респондентів за допомогою .head(5)
    df_top5 = df_clean_comp.sort_values(by=comp_col, ascending=False).head(5)

    #Формуємо фінальну таблицю, залишаючи лише компенсацію та рівень освіти
    df_final_result = df_top5[[comp_col, edu_col]].reset_index(drop=True)

    # Виведення результатів у зручному форматі таблиці
    print("Рівні освіти 5 респондентів з найбільшою компенсацією:")
    display(df_final_result)

else:
    print(f"Помилка! Перевірте наявність колонок '{comp_col}' та '{edu_col}' в датасеті.")
```

```python
#Завдання 9. Аналіз популярності Python по віковим категоріям
#Визначаємо назви колонок
age_col = 'Age'
lang_col = 'LanguageHaveWorkedWith'

# Перевіряємо наявність обох колонок у датасеті
if age_col in df_results.columns and lang_col in df_results.columns:

    # Очищуємо дані: беремо лише тих, у кого вказано і вік, і мови (без NaN)
    df_filtered = df_results[[age_col, lang_col]].dropna()

    # Створюємо булеву колонку: True, якщо респондент знає Python, і False, якщо ні
    df_filtered['Uses_Python'] = df_filtered[lang_col].str.contains('Python', na=False)

    # Групуємо за віковими категоріями та обчислюємо відсоток
    # Оскільки True рахується як 1, а False як 0, середнє значення (mean)
    # автоматично дасть нам частку користувачів Python у кожній групі
    age_analysis = df_filtered.groupby(age_col)['Uses_Python'].mean() * 100

    # 3. Перетворюємо результат у гарний DataFrame
    df_age_result = age_analysis.reset_index()
    df_age_result.columns = ['Вікова категорія', 'Відсоток Python-розробників']

    # Округляємо відсотки до 1 знака після коми для чіткості
    df_age_result['Відсоток Python-розробників'] = df_age_result['Відсоток Python-розробників'].round(1)

    # Форматуємо вивід, щоб додати значок %
    df_age_result['Відсоток Python-розробників'] = df_age_result['Відсоток Python-розробників'].astype(str) + '%'

    #Виведення фінального результату
    print("Популярність мови Python за віковими категоріями респондентів:")
    display(df_age_result)

else:
    print(f"Помилка! Перевірте наявність колонок '{age_col}' та '{lang_col}' в датасеті.")
#Можемо говорити про таку тенденцію: Пайтон найбільш популярний серед респондентів молодшого віку.
```

```python
#Завдання 10. Аналіз індустрій серед високооплачуваних віддалених працівників

#Визначаємо назви колонок
comp_col = 'ConvertedCompYearly'
remote_col = 'RemoteWork'
industry_col = 'Industry'

# Перевіряємо наявність усіх необхідних колонок в датасеті
required_cols = [comp_col, remote_col, industry_col]
if all(col in df_results.columns for col in required_cols):

    # Очищуємо дані від NaN у колонці компенсації для коректного розрахунку перцентиля
    df_valid_comp = df_results[df_results[comp_col].notna()]

    # 2. Розраховуємо 75-й перцентиль (квантиль 0.75) компенсації
    quantile_75 = df_valid_comp[comp_col].quantile(0.75)
    print(f"Порогове значення 75-го перцентиля компенсації: ${quantile_75:,.2f} на рік")
    print("Це означає, що 25% найоплачуваніших респондентів заробляють більше цієї суми.\n")

    # 3. Виконуємо множинну фільтрацію за двома критеріями:
    # Критерій А: Компенсація >= 75 перцентиль
    # Критерій Б: Працюють віддалено (містить слово 'Remote')
    high_earning_remote = df_valid_comp[
        (df_valid_comp[comp_col] >= quantile_75) &
        (df_valid_comp[remote_col] == 'Remote')
    ]

    #Визначаємо найрозповсюдженіші індустрії та рахуємо їх частоту
    #Метод .value_counts() автоматично сортує результати за частотою від більшої до меншої
    industry_counts = high_earning_remote[industry_col].dropna().value_counts()

    # Перетворюємо в DataFrame для красивого представлення
    df_industry_result = industry_counts.reset_index()
    df_industry_result.columns = ['Індустрія', 'Кількість респондентів']

    # 5. Виведення фінального результату
    print("Найрозповсюдженіші індустрії серед високооплачуваних віддалених працівників:")
    display(df_industry_result)

else:
    print("Помилка! Перевірте наявність усіх колонок у датасеті:", required_cols)
```
